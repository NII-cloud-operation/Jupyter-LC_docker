# OperationHubをAWSに構築

[OperationHub](https://github.com/NII-cloud-operation/OperationHub) は、チームの運用者がJupyter Notebookを運用管理端末として利用するためにNIIがカスタマイズしたJupyterHubです。一般的に運用管理端末としてWindows端末などを運用チームからアクセス可能にすることがあると思いますが、この役割をJupyter Notebookで行おうというものです。

![operationhub-architecture](./images/operationhub-architecture.png)

Notebook [05_OperationHubをAWSに構築](05_OperationHubをAWSに構築.ipynb) を使うと、AWS上の仮想マシンとしてOperationHubを立ち上げることができます。

このNotebookの実行には**AWSアカウント**が必要です。ご自身の責任でNotebookをご利用ください。
